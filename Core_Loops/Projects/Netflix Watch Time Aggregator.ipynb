```python
watch_time = [40, 20, 55, 100, 10]

total = 0
largest = 0

for w in watch_time:
    total += w
    if w > largest:
        largest = w

avg = total / len(watch_time)

print(f"Total watch time: {total} minutes")
print(f"Average watch time: {avg} minutes")
print(f"Longest watch time: {largest} minutes")
```
