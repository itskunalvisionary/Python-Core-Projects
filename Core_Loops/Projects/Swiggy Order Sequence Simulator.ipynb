```python
veg = 0
nonveg = 0
highest = 0

while True:
    item = input("Item name (or done): ")

    if item == "done":
        break

    price = int(input("Price: "))
    type_ = input("veg/nonveg: ")

    if type_ == "veg":
        veg += 1
    else:
        nonveg += 1

    if price > highest:
        highest = price
    
print("Veg:", veg)
print("Non-Veg:", nonveg)
print("Highest Price:", highest)
```
