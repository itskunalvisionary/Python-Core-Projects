```python
# Q1. Print numbers 1 to 100

for i in range(1,101):
    print(i)
```

```python
# Q2. Multiplication Table

n =5
for i in range(1,11):
    print(f"{n} X {i} = {n*1}")
```

```python
# Q3. Count digits of a number
n = 123456
count = 0

while n > 0:
    count += 1
    n //= 10

print(count)
```

```python
# Q4. Sum of first N natural numbers

N = 10
total = 0

for i in range(1, N+1):
    total += i

print(total)
```

```python
# Q5. Reverse a number using while

n = 12345
rev = 0
while n >0:
    digit = n%10
    rev = rev *10 + digit
    n //= 10

print(rev)


# ------------------------------------------
# Q5. Reverse a number using string slicing
n = 12345
a = str(n)[::-1]
print(a)
```

```python
# Q6. Factorial

n = 5
fact = 1

for i in range (1, n+1):
    fact *= i
print(fact)
```

```python
# Q7. Print even numbers

for i in range(1,51):
    if i % 2 == 0:
        print(i)
```

```python
# Q8. Skip numbers divisible by 5

for i in range(1,51):
    if i % 5 == 0:
        continue
    print(i)
```

```python
# Q9. Sum of digits

n = 1234
total = 0

for ch in str(n):
    total += int(ch)
print(total)
```
