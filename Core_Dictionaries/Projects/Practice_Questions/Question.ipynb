```python
# Q1. Create a dictionary and print values

student = {
    "Name" : "Kunal",
    "Age" : 17,
    "Marks" : 65
}

for Key in student:
    print(student[Key])
```

```python
# Q7. Count frequency of characters

word = "banana"
freq = {}

for ch in word:
    if ch not in freq:
        freq[ch] = 1
    else:
        freq[ch] += 1

print(freq)
```

```python
# Q8. Merge two dictionaries

a = {"x" : 1}
b = {"y" : 2}

merged = {**a,**b}
print(merged)
```

```python
# Q9. Find student with highest marks (list of dicts)

data = [
    {"name":"A","marks":50},
    {"name":"B","marks":90},
    {"name":"C","marks":75}
]

best = data[0]

for d in data:
    if d["marks"] > best["marks"]:
        best = d
print(best)
```

```python
# Q10. Convert 2 lists into dictionary

keys = ["Name", "Age"]
value = ["Kunal",17]

d = {}

for i in range(len(keys)):
    d[keys[i]] = value[i]
print(d)
```
