```python
"""  
Goal:
✔ create dict of video → stats
✔ find highest viewed
✔ calculate CTR
✔ print clean report

"""
```

```python
titles = ["Video1","Video2","Video3"]
views  = [100,200,50]
likes  = [10,20,5]

yt = {}

for i in range(len(titles)):
    yt[titles[i]] = {
        "views": views[i],
        "likes": likes[i]
    }

print(yt)

# Find highest viewed
best = None
for name, data in yt.items():
    if best is None or data["views"] > yt[best]["views"]:
        best = name

print("Top Video:", best)
```
