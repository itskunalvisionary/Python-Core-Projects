```python
"""   
Goal:
✔ find total watch time
✔ most watched show
✔ add new show
✔ remove show
✔ update timestamp

"""
```

```python
user = {
  "name":"Kunal",
  "watch_history":[
      {"title":"Money Heist","time":120},
      {"title":"Suits","time":90}
  ]
}

#add show
user["watch_history"].append({"title":"Friends","time":150})

#total time

total = 0
best = None

for show in user["watch_history"]:
    total += show["time"]
    if best is None or show["time"] > best["time"]:
        best = show

print("Total Watch Time:", total)
print("Most Watched:", best["title"])
```
