```python
"""  
Goal:
✔ count each word
✔ find most used
✔ store in dict
✔ print sorted result

"""
```

```python
comment = "wow awesome awesome cool wow wow nice"
words = comment.split()

freq = {}

for w in words:
    freq[w] = freq.get(w, 0) + 1

print(freq)

most = max(freq, key=freq.get)
print("Most Used Word:", most)
```
