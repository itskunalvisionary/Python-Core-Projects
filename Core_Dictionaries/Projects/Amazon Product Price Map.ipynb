```python
"""  
Goal:
✔ convert into dict
✔ find cheapest
✔ find average
✔ show price chart

"""
```

```python
products = ["Laptop","Mouse","Keyboard"]
prices = [50000,500,800]

d = { products[i]: prices[i] for i in range(len(products))}

cheapest = min(d,key=d.get)
avg = sum(prices)/len(prices)


print("Map:", d)
print("Cheapest:", cheapest)
print("Average:", avg)
```
