```python
# Q1. Create a tuple and print elements

t = (10,20,30,40)

for x in t:
    print(x)
```

```python
# Q2. Try to modify tuple (should fail)

t = (1,2,3,4)
t[0] = 11
```

```python
# Q3. Convert list → tuple

lst = [1,2,3,4,5]
print(type(lst))
t = tuple(lst)
print(type(t))
```

```python
# Q4. Find length of tuple

t = 10,20,30,40
print(len(t))
```

```python
# Q5. Create a set & observe duplicate removal

s = {1,2,2,3,3,4,5,5}
print(s)
```

```python
# Q6. Add item to set

s = {1,2,3,4,5,6}
s.add(7)
print(s)
```

```python
# Q7. Remove item from set

s = {1,2,3,4,5,6,7,8,9}
s.remove(9)
print(s)
```

```python
# Q8. Check membership in set

print(5 in {1,2,3})
```

```python
# Q9. Union of sets

a = {1,2}
b = {2,3}
print(a | b)
```

```python
# Q10. Intersection of sets

a = {1,2,3}
b = {2,3,4}
print(a & b)
```
