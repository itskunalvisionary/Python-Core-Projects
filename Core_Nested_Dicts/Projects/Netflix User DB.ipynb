```python
"""
Goal:

✔ Total watch time
✔ Most active profile
✔ Add new profile
✔ Add watch-time

"""
```

```python
user = {
    "name":"Kunal",
    "profiles":[
        {"name":"Kids","watch":[30,50]},
        {"name":"Main","watch":[120,90]}
    ]
}

#add new profile and watch-time

user["profiles"].append({"name" : "Anime", "watch" : [60]})

#Total Watch & Best 

total = 0
best = None
best_time = 0

for profile in user["profiles"]:
    t = sum(profile["watch"])
    total += t
    if t > best_time:
        best_time = t
        best = profile["name"]

print("Total Watch:", total)
print("Top Profile:", best)

```
