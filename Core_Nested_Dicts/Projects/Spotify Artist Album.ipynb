```python
"""  
Goals:

✔ Total streams
✔ Most streamed song
✔ Add new song
✔ Add new artist
✔ Update streams

"""
```

```python
spotify = {
 "Arijit": {
     "Tum Hi Ho": {"year":2013, "streams":200},
     "Kesariya": {"year":2022, "streams":500}
 }
}

# add new artist

spotify["AP Dhillon"] = {
    "Excuses" : {"year" : 2020, "streams" : 350}
}

# Add song to existing artist
spotify["Arijit"]["gerua"] = {"year" : 2015, "streams" : 300}

# Find most streamed
top_song = None
top_streams = 0

total = 0

for artist,song in spotify.items():
    for title,info in song.items():
        total += info["streams"]
        if info["streams"] > top_streams:
            top_streams = info["streams"]
            top_song = title

print("Total Streams:", total)
print("Most Streamed:", top_song)
```
