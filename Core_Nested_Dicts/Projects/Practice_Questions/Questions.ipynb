```python
# Q1. Access values from a nested dictionary

user = {
    "name": "Kunal",
    "details": {
        "city": "Mumbai",
        "age": 20
    }
}

print(user["details"]["city"])
print(user["details"]["age"])
```

```python
# Q2. Add new nested keys

user["details"]["pincode"] = 400001
```

```python
user
```

```python
# Q3. Loop through nested dict

for k,v in user["details"].items():
    print(f"{k} - {v}")
```

```python
# Q4. Nested list inside dict

student = {
    "name": "Kunal",
    "scores": [90, 85, 70]
}

print(student["scores"][1])
print(student["scores"][0])
```

```python
# Q5. Add dict inside list

student["scores"].append(95)
print(student)

```

```python
# Q6. Store multiple users inside dictionary

user = {"u1":{},
        "u2" : {}}

user["u1"]["Name"] = "Kunal"
user["u1"]["Age"] = 17
user["u2"]["Name"] = "Vikas"
user["u2"]["Age"] = 23

user
```

```python
# Q7. Update nested dict values

user["u1"]["Age"] = 18

print(user["u1"])
```

```python
# Q8. Delete nested key
del user["u2"]["Age"]
```

```python
user
```

```python
# Q9. Count total keys inside nested dict

count = 0

for d in user.items():
    count += len(d)

print(count)
```

```python
# Q10. Convert nested dict to flat dict

a = {"name":"Kunal","details":{"age":20,"city":"Mumbai"}}


flat = {}
for k,v in a.items():
    if isinstance(v, dict):
        for k2,v2 in v.items():
            flat[k2] = v2
    else:
        flat[k] = v

print(flat)
```
