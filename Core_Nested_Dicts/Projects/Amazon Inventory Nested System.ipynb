```python
inv = {
 "electronics": {
     "laptop": {"price":50000, "stock":10},
     "mouse": {"price":500, "stock":50}
 },
 "fashion": {
     "shirt": {"price":700, "stock":100}
 }
}

# add
inv["fashion"]["jeans"] = {"price" : 1200, "stock" : 40}

max_item = None
max_price = 0
total_stock = 0

for k,v in inv.items():
    for k2,v2 in v.items():
        if v2["price"] > max_price:
            max_price = v2["price"]
            max_item = k2
        total_stock += v2["stock"]

print("Most Expensive:", max_item)
print("Total Stock:", total_stock)       

```
