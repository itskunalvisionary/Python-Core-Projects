Input: random watch-times

Goal:

✔ Sort by watch-time

✔ Top 10 display

✔ Remove duplicates

✔ Clean negative values

```python
watch = [50,120,200,200,300,500,20,100,400,300,150]

clean = [x for x in watch if x >= 0]
unique = []

for x in clean:
    if x not in unique:
        unique.append(x)

top10 = sorted(unique, reverse=True)[:10]
print(top10)
```
