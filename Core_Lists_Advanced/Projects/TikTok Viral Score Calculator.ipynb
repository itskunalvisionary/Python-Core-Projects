Score formula:

score = likes*0.5 + comments*1 + shares*2

```python
videos = [
    [100, 20, 10],   # likes, comments, shares
    [300, 50, 40],
    [50, 5, 2]
]

scores = []

for like, com, share in videos:
    score = like*0.5 + com*1 + share*2
    scores.append(score)

print(scores)
print("Most Viral Score:", max(scores))   
```
