Goal:

✔ flatten
✔ filter
✔ map
✔ compress nested data
✔ produce final one-line list

```python
mat = [[1,2,3],[4,5,6],[7,8,9]]

flat = [num for x in mat for num in x]
filter = [x for x in flat if x%2 == 0]
squares = [x*x for x in filter]

print("Flat:", flat)
print("Filtered Evens:", filter)
print("Squares:", squares)
```
