```python
# Q1. Remove duplicates from a list (WITHOUT set())

lst = [1,2,2,3,3,3,4]
unique = []

for x in lst:
    if x not in unique:       #👉 “Not in = gatekeeper check.”
        unique.append(x)

print(unique)
```

```python
# Q2. Sort list manually (WITHOUT sort())

lst = [5,1,4,2]
for i in range(len(lst)):
    for j in range(i+1, len(lst)):
        if lst[j] < lst[i]:
            lst[i], lst[j] = lst[j], lst[i]

print(lst)

```

```python
# Q3. Flatten a nested list

nested = [[1,2],[3,4],[5]]
flat = []

for group in nested:
    for num in group:
        flat.append(num)
print(flat)
```

```python
# Q4. Find the second largest number

lst = [10, 5, 20, 8]

mx = max(lst)
lst.remove(mx)
print(max(lst))
```

```python
# Q5. Filter even numbers using list comprehension

lst = [1,2,3,4,5,6]
evens = [x for x in lst if x%2 == 0]
print(evens)
```

```python
# Q6. Square each element (list comprehension)

nums = [1,2,3]
sq = [x*x for x in nums]
print(sq)
```

```python
# Q7. Reverse list using slicing

lst = [1,2,3,4]
rev = lst[::-1]
print(rev)
```

```python
# Q8. Count frequency of elements

lst = [1,2,2,3,3,3]
freq = {}

for x in lst:
    if x not in freq:
        freq[x] = 1
    else:
        freq[x] += 1
print(freq)
```

```python
# Q9. Pair items of two lists

a = [1,2,3]
b = ["a","b","c"]

pairs = []
for i in range(len(a)):
    pairs.append((a[i],b[i]))

pairs
```

```python
# Q10. Remove all zeros

lst = [0,1,0,2,3,0,4]

clean = [x for x in lst if x != 0]
print(clean)
```
