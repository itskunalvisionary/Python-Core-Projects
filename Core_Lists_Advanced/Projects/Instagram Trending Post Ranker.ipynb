Goals:

✔ Sort descending

✔ Top 3 posts

✔ Average views

✔ Remove low (<100)

```python
posts = [120, 300, 250, 700, 50]

sorted_posts = sorted(posts, reverse=True)
top3 = sorted_posts[:3]
avg = sum(posts)/len(posts)
filtered = [p for p in posts if p >= 100]

print("Sorted:", sorted_posts)
print("Top 3:", top3)
print("Average:", avg)
print("Filtered:", filtered)
```
