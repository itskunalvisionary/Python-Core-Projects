Goal:
✔ print all valid (row, column) positions of a Knight (horse)
✔ from given position (r,c)

```python
moves = [(2,1),(2,-1),(-2,1),(-2,-1),(1,2),(1,-2),(-1,2),(-1,-2)]

r,c = 4, 4 # Board size

vaild = []

for dr, dc in moves:
    nr = r + dr
    nc = c + dc
    if 1 <= nr <= 8 and 1 <= nc <= 8:
        vaild.append((nr,nc))
print(vaild)
```
