Input:

items cooked per hour

rows = hours

columns = item IDs

Goal:
✔ find max cooked item
✔ total cooking
✔ find hours with lowest output

```python
kitchen = [
    [10,20,15],
    [5,30,25],
    [12,18,22]
]

total = 0
max_item = 0

for row in kitchen:
    for item in row:
        total += item
        if item > max_item:
            max_item = item

print("Total items in kitchen:", total)
print("Max item quantity:", max_item)
```
