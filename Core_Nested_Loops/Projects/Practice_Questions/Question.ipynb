```python
# Q1. Print a Square Pattern (5x5)

for i in range(5):
    print("*" * 5)
```

```python
# Q2. Right Triangle Pattern

for i in range(1,6):
    print("*" * i)
```

```python
# Q3. Reverse Triangle Pattern

for i in range(5,0,-1):
    print("*"*i)
```

```python
# Q4. Print Table from 1 to 4 (Nested Loop)

for i in range(1,5):
    for j in range(1,11):
        print(f"{i} x {j} = {i*j}")
    print()  # Print a new line after each table
```

```python
# Q5. Number Pyramid

for i in range(1,6):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
# Q6. Counting Occurrences in Nested List

lst = [[1,2,3],[4,5,1],[1,8,9]]
count = 0

for rows in lst:
    for num in rows:
        if num == 1:
            count += 1

print("The number 1 occurs", count, "times")
```

```python
# Q7. Print Chessboard Pattern (8x8)

for i in range(8):
    for j in range(8):
        if (i + j) % 2 == 0:
            print("#", end="")
        else:
            print(".", end="")
    print()
```

```python
# Q8. Find Max Element in Nested List

matrix = [[10,20,3],[4,99,1],[2,5,6]]

mx = matrix[0][0]

for row in matrix:
    for num in row:
        if num > mx:
            mx = num
print("The maximum element is:", mx)
```

```python
# Q9. Count Total Characters in 2D String List

data = [["hi","hello"],["python","rocks"]]

total_chars = 0

for row in data:
    for string in row:
        total_chars += len(string)
print("Total characters:", total_chars)
```

```python
# Q10. Matrix Addition (2x2)

A = [[1,2],[3,4]]
B = [[5,6],[7,8]]

result = [[0,0],[0,0]]

for i in range(2):
    for j in range(2):
        result[i][j] = A[i][j] + B[i][j]

print(result)
```
